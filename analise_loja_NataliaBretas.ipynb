##6. Gráficos

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar os dados das lojas
url1 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/main/base-de-dados-challenge-1/loja_4.csv"

loja1 = pd.read_csv(url1)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

# Calcular faturamento total por loja
faturamentos = {
    'Loja 1': int(loja1['Preço'].sum()),
    'Loja 2': int(loja2['Preço'].sum()),
    'Loja 3': int(loja3['Preço'].sum()),
    'Loja 4': int(loja4['Preço'].sum())
}

# Criar DataFrame para visualização
df_faturamento = pd.DataFrame(list(faturamentos.items()), columns=['Loja', 'Faturamento'])

# Estilo do gráfico
sns.set_style("whitegrid")
plt.figure(figsize=(10, 6))
grafico = sns.barplot(x='Loja', y='Faturamento', data=df_faturamento, palette='crest')

# Adicionar valores inteiros sobre as barras
for i, valor in enumerate(df_faturamento['Faturamento']):
    grafico.text(i, valor + 10000, f'R$ {valor:,}', ha='center', va='bottom', fontsize=12, fontweight='bold')

# Título e rótulos
plt.title('Faturamento Total por Loja (R$)', fontsize=16, fontweight='bold')
plt.ylabel('Faturamento (R$)', fontsize=12)
plt.xlabel('')
plt.tight_layout()
plt.show()
```

```python
# Unir todos os dados
todas_lojas = pd.concat([loja1, loja2, loja3, loja4])

# Contar categorias
categorias = todas_lojas['Categoria do Produto'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(categorias, labels=categorias.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribuição das categorias de produtos')
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(todas_lojas['Preço'], todas_lojas['Frete'], alpha=0.6, color='royalblue')
plt.title('Relação entre preço e frete')
plt.xlabel('Preço (R$)')
plt.ylabel('Frete (R$)')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# URLs dos dados
url1 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/main/base-de-dados-challenge-1/loja_4.csv"

# Carregar os dados
loja1 = pd.read_csv(url1)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

# Calcular custo médio de frete
fretes = {
    'Loja 1': loja1['Frete'].mean(),
    'Loja 2': loja2['Frete'].mean(),
    'Loja 3': loja3['Frete'].mean(),
    'Loja 4': loja4['Frete'].mean()
}

# Preparar dados para gráfico
lojas = list(fretes.keys())
valores = list(fretes.values())

# Criar gráfico de linha
plt.figure(figsize=(8, 5))
plt.plot(lojas, valores, marker='o', linestyle='-', color='darkorange', linewidth=2)
for i, valor in enumerate(valores):
    plt.text(lojas[i], valor + 0.3, f'R$ {valor:.2f}', ha='center', fontsize=11)

plt.title('📦 Custo Médio de Frete por Loja', fontsize=16, fontweight='bold')
plt.ylabel('Valor médio (R$)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
